# ASPIRE Run Notebook

This notebook is the short run companion for the ASPIRE pipeline. The canonical pipeline documentation is `README.md`.

## Current Scope

ASPIRE runs the maintained Nextflow workflow in `asv_pipeline.nf`: FASTQ QC, merge/filter, ASV inference, taxonomy, optional mitochondrial/contaminant filtering, metadata-linked ASV summaries, ecological analyses, VOC correlation, SPIEC-EASI/network modules, optional patient-aware analyses, ASV-to-MAG linkage, Sankey summaries, and master summary export.

## 1. Prepare A Config

Start from `asv_pipeline_nextflow.yml` for a new run, or use `examples/set1-2.local.yml` only for the local UBC/LMP test dataset. Edit all input, output, metadata, reference, VOC, and optional branch paths before running.

```python
cp asv_pipeline_nextflow.yml my_run.yml
```

## 2. Run The Pipeline

Use the wrapper for routine runs. It bootstraps the controller environment, sets work/cache paths from YAML, and launches Nextflow with the correct params file.

```python
./run_asv_pipeline.sh my_run.yml
```

## 3. Rerun A Stage

If Nextflow cache prevents a branch from rerunning after a config or script change, use `--rerun-from` with a stage name from `./run_asv_pipeline.sh --list-stages`.

```python
./run_asv_pipeline.sh my_run.yml --rerun-from VOC_CORRELATION
```

## 4. Important Outputs

- `ASVs/ASV_target.tsv`: final filtered microbial ASV count table used downstream.
- `metadata/`: metadata-linked ASV tables and plots.
- `voc_correlation/`: ASV-VOC correlation outputs and VOC abundance z-score plots when enabled.
- `spieceasi/` and `network/`: network and module outputs when enabled.
- `master_summary/`: combined final summary outputs when enabled.
