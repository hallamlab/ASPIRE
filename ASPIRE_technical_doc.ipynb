# ASPIRE Technical Documentation
This notebook documents the current repo entrypoints and major scripts that remain part of the active ASPIRE workflow.

## Scope
- The legacy biochemistry preprocessing branch has been removed from the repository.
- This document now focuses on the maintained ASV pipeline and downstream components that are still callable from `asv_pipeline.nf`.

## run_asv_pipeline.sh
- Purpose: bootstrap the controller environment and launch the current Nextflow workflow.
- Inputs: `asv_pipeline_nextflow.yml` (or another params file) plus `processes/controller/env.yml`.
- Behavior: resolves work/cache paths, manages resume behavior, and runs `nextflow run asv_pipeline.nf`.

## asv_pipeline.nf
- Purpose: orchestrate the maintained ASV workflow from raw reads through downstream analysis branches.
- Core stages: `FASTP_QC`, `MERGE_READS`, `FILTER_READS`, `CONCAT`, `DEREPLICATE`, `SINA_TRIM`, `DENOISE`, `CHIMERA_CHECK`, `CREATE_COUNT_MATRIX`, `FILTER_TABLE`.
- Optional maintained branches: metadata plots, batch correction, outlier checking, collectors curves, upset plots, UMAP clustering, diversity, indicator species, clustermaps, SpiecEasi/network analysis, power analysis, patient-aware analyses, lung-status analysis, ASV-MAG linking, sankey summaries, and master summary export.

## processes/plot_metadata/plot_metadata.py
- Purpose: build the microbial and mitochondrial metadata-derived tables used by several downstream branches.
- Inputs: metadata table, taxonomy table, sample manifest, ASV count tables, and optional auxiliary metadata merges.
- Outputs: updated metadata plus `ASV_meta_*` and `ASV_final.*` tables.

## processes/umap_clustering/umap_clustering.py
- Purpose: create maintained UMAP/HDBSCAN clustering outputs from ASV-linked metadata tables.
- Notes: this script is still referenced by `asv_pipeline.nf` through the `umap_clustering` optional branch.

## Lung-status components
- `processes/lung_status_analysis/prepare_lung_status_data.py`: derives the per-sample lung-status labels and filtered input tables.
- `processes/lung_status_analysis/run_lung_status_analysis.R`: runs the statistical comparisons.
- `processes/lung_status_analysis/plot_lung_status_analysis.py`: produces the companion figures.
- These scripts are still active as the maintained `lung_status_analysis` branch.

## Network and summary components
- `processes/spieceasi/run_spieceasi.R`, `processes/graph_network/graph_network.py`, `processes/network_modules/network_modules.R`, and `processes/master_summary/build_master_asv_summary.py` remain part of the active downstream analysis path.
- `processes/power_analysis_pipeline/run_power_analysis_pipeline.sh` remains the maintained wrapper for the power-analysis branch.
